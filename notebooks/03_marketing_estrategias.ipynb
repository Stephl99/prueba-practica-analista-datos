4. Una marca acaba de realizar una activación por medio de 3 posts en Facebook. Los siguientes son los resultados de la campaña:

**Pieza 1:**
Impresiones: 5.200
Alcance: 1.450
Interacciones: 16

**Pieza 2:**
Impresiones: 2.500
Alcance: 800
Interacciones: 67

**Pieza 3:**
Impresiones: 730
Alcance: 150
Interacciones: 93

**a.¿Cuál es el alcance total de la campaña?**

$Alcance = 1450 + 800 + 150 = 2400$

b.¿Considera que la frecuencia es apropiada?

La frecuencia promedio de la campaña fue de 3.51, es decir que una persona
alcanzada vio el contenido alrededor de 3.5 veces. A nivel general, es una frecuencia razonable, pero dependiendo de lo que se busque con la publicidad podría cambiar la reocmendación. Si se busca recordación o interacción de las personas, aumentar la frecuencía podría ser favorable, mientras que si se quiere ampliar el alcance de la publicidad, se podría sugerir una frecuencia menor.

c.Si nuestro cliente es conocido en el mercado y lleva ya varios años
posicionado en redes sociales, ¿cuál de las piezas funcionó mejor? y
bajo ¿qué criterio puede comparar los resultados de cada una?

Como el cliente es conocido y lleva varios años posicionado en redes sociales, el objetivo de la campaña dbee ir enfocado en interacción y conexión con las personas. Con esto en mente, se considera la tasa de interacción como el criterio para comparar las piezas publicitarias.

$Engagement = Interacciones / Alcance$

Se observa que la pieza 3 tuvo una mejor tasa de interacción que las demás con bastante margen de diferencia. Esto nos muestra que el 62% de las personas que vieron esta pieza, realmente hicieron alguna interacción, que es lo que se busca para el caso.

```python
import pandas as pd
```

```python
data = pd.DataFrame({'Pieza': [1,2,3],
        'Impresiones': [5200,2500,730], 
        'Alcance': [1450,800,150], 
        'Interacciones': [16,67,93]})

# a
data['Alcance'].sum()

# b
data['Impresiones'].sum() / data['Alcance'].sum()

# c
data['engagement_rate'] = data['Interacciones'] / data['Alcance']
```
